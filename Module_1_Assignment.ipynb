```python
# Q6
hp = df['horsepower']

# [Step 1] median value of horsepower (with missing values)
med_hp = hp.median()
print(f'Median value of horsepower (with missing values):\n {med_hp}')

# [Step 2] most frequent value of horsepower (with missing values)
mfv_hp = hp.mode()
print(f'\nMost frequent value of horsepower (with missing values):\n {mfv_hp}')

# [Step 3] fill missing values with most frequent value
hp.fillna(mfv_hp, inplace=True)

# [Step 4] median value of horsepower (after filling missing values)
med_hp = hp.median()
print(f'\nMedian value of horsepower (after filling missing values):\n {med_hp}')
```

```python
# Q7
# [Step 1] Select all cars from Asia
asia = df[df['origin'] == 'Asia']

asia.head()
```

```python
# [Step 2] Select only vehicle_weight and model_year columns
asia_df = asia[['vehicle_weight', 'model_year']]

asia_df.head()
```

```python
# [Step 3] Select first 7 values
asia_7 = asia_df.iloc[:7]

asia_7.head(7)
```

```python
# [Step 4] Get the underlying numpy array
X = asia_7.to_numpy()

X
```

```python
# [Step 5a] Transpose X
XT = X.T

XT
```

```python
# [Step 5b] matrix-matrix multiplication between X and XT
XTX = np.dot(XT, X)

XTX
```

```python
# [Step 6] Invert XTX
XTX_inv = np.linalg.inv(XTX)

XTX_inv
```

```python
# [Step 7] Create array y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

y
```

```python
# [Step 8] Multiply the inverse of XTX with the transpose of X, and then multiply the result by y
w = np.dot(np.dot(XTX_inv, XT), y)

w
```

```python
# [Step 9] Sum of w
w.sum()
```
